```python
import pandas as pd

df1 = pd.DataFrame({
    "A": ["ΑΘ", "A1", "A2", "A3"], 
    "B": ["80", "81", "82", "83"], 
    "C": ["CO", "C1", "C2", "C3"], 
    "D": ["DO", "D1", "D2", "D3"], 
 }, index=[0,1,2,3])

df2 = pd.DataFrame({
    "A": ["A4", "AS", "A6", "A7"],
    "B": ["BA", "B", "86","87"], 
    "C": ["CA", "CS", "06", "C7"], 
    "D": ["04","05", "06", "07"], 
 }, index=[4,5,6,7])

df3 = pd.DataFrame({
    "A": ["AB", "AS", "A10", "A11"], 
    "B": ["88","89","810", "811"],
    "C": ["CB", "C", "C10", "C11"], 
    "D": ["08","09", "010", "D11"], 
 }, index=[8,9,10,11])
```

```python
print(df1)
```

```python
print(df2)
```

```python
print(df3)
```

```python
resultado = pd.concat([df1,df2], axis = 1)
print(resultado)
```

```python
frames = [df1, df2, df3]
result = pd.concat(frames)
result
```

```python
resultado = pd.concat([df1, df2], axis = 1, join = 'outer')
resultado
```

```python
resultado = pd.concat([df1, df3], axis = 1, join = 'inner')
resultado
```
