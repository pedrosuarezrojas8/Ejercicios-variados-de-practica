```python
import numpy as np
import pandas as pd
```

```python
# Creamos un generador de números aleatorios utilizando la función default_rng() de NumPy.
rng = np.random.default_rng()

#Usamos para generar números enteros aleatorios entre 0 y 100, con un tamaño de 10.
a_values = rng.integers(low = 0, high = 100, size = 10) 

#Usamos para generar números enteros aleatorios entre 100 y 500, con un tamaño de 10.
b_values = rng.integers(low = 100, high = 500, size = 10)

#Creamos un DataFrame de pandas con dos columnas, 'A' y 'B', utilizando los valores generados anteriormente.
data = pd.DataFrame({'A': a_values, 'B': b_values})
data
```

```python
# Ahora con cut podemos categorizar los valores de la columna 'A' en diferentes rangos, asignando etiquetas a cada rango.

data['A_cat'] = pd.cut(data.A, bins = [0, 20, 40, 60, 80, 100], labels = ['Muy poco', 'Poco', 'Normal', 'Mucho', 'Demasiado'])
data
```

```python
# Aqui creamos un diccionario para MAPEAR las categorias

A_cat_mapper = {'Poco': 'Insuficiente',
                'Muy poco' : 'Insuficiente',
                'Normal' : 'Suficiente',
                'Mucho' : 'Suficiente',
                'Demasiado' : 'Suficiente'}
```

```python
data['A_cat_mapper_insuf'] = data.A_cat.map(A_cat_mapper)
data
```

```python
# Usamos dummies para convertir las categorias de la columna 'A_cat' en variables dummy (variables binarias) para cada categoria.

A_categories_1_dummies = pd.get_dummies(data.A_cat, drop_first = True, prefix = 'A_cat1', dtype = int)
A_categories_1_dummies
```

```python
A_categories_2_dummies = pd.get_dummies(data.A_cat_mapper_insuf, drop_first = True, prefix = 'A_cat2', dtype = int)
A_categories_2_dummies
```
