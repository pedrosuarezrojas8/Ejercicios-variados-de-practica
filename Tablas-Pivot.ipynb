```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset('penguins')
print(df.shape)
df.sample(5)
```

```python
df.pivot_table('body_mass_g', index = 'species', columns = 'sex')
```

```python
df['flipper_length_mm'].describe()[['min', 'mean', 'max']]
```

```python
flipper_cat = pd.cut(df.flipper_length_mm, [172, 200, 231])
df.pivot_table('body_mass_g',['species', flipper_cat,],'sex')
```

```python
bill_categories = pd.qcut(df.bill_length_mm, 3, labels = ['pequeño', 'mediano', 'grande'])
df.pivot_table('body_mass_g', ['species', flipper_cat],[bill_categories, 'sex'], fill_value = "No existe")
```
