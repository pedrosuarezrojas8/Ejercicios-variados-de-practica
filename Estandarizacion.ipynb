```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

NumPy → Librería para trabajar con arrays y operaciones matemáticas.
Pandas → Manejo de datos en DataFrames.
load_iris → Carga el dataset Iris incluido en sklearn.
train_test_split → Divide los datos en entrenamiento y prueba.
StandardScaler → Estandariza los datos (media 0, varianza 1).
RandomForestClassifier → Modelo de clasificación basado en árboles.
accuracy_score → Métrica para evaluar el modelo.

```python
iris = load_iris()
x = iris.data #data es la matriz de caracteristicas
y = iris.target #Target es la variable que queremos predecir

df = pd.DataFrame(x, columns = iris.feature_names)
df['target'] = y
print(df.head())
```

Carga el dataset Iris como un objeto tipo Bunch (similar a un diccionario).

x → matriz de características (150 filas × 4 columnas).
y → vector con las clases (0, 1, 2).

Convierte los datos en un DataFrame.
Añade la columna target.
df.head() muestra las primeras 5 filas.


```python
x_train, x_test, y_train, y_test = train_test_split(x, y , test_size = 0.2, random_state = 42)
```

train_test_split divide los datos:

80% entrenamiento
20% prueba
random_state=42 asegura reproducibilidad.

```python
scaler = StandardScaler()
x_train_standardized = scaler.fit_transform(x_train)
x_test_standardized = scaler.transform(x_test)
```

Crea un objeto escalador.

fit() calcula media y desviación estándar del train.
transform() aplica la transformación.
fit_transform() hace ambas cosas en una sola línea.
x_test_standardized = scaler.transform(x_test)

```python
clf = RandomForestClassifier(random_state= 42)
clf.fit(x_train_standardized, y_train)

y_pred_standardized = clf.predict(x_test_standardized)

accuracy_standardized = accuracy_score(y_test, y_pred_standardized)
print(f'Precisión con datos estandarizados: {accuracy_standardized:.2f}')
```

RandomForestClassifier crea un modelo basado en muchos árboles de decisión.

random_state=42 asegura que los resultados sean reproducibles.

predict() toma los datos de prueba estandarizados y predice la clase de cada muestra.

El modelo no ve estos datos durante el entrenamiento, por lo que esta predicción es válida para evaluar rendimiento real.

accuracy_score compara:

y_test → valores reales

y_pred_standardized → valores predichos

Devuelve un número entre 0 y 1:

1.00 = 100% de acierto
0.90 = 90% de acierto

```python
clf.fit(x_train, y_train)
y_pred = clf.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

Entrenas el modelo Random Forest usando:

x_train → datos originales (sin escalar)
y_train → etiquetas reales

El modelo aprende patrones directamente de los valores crudos.
El modelo predice las clases del conjunto de prueba.
Usa los datos sin estandarizar, tal como fueron cargados.

Compara:

y_test → valores reales
y_pred → predicciones del modelo

Devuelve un valor entre 0 y 1 que indica qué tan bien predijo.
