```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
```

```python
# Creamos un generador de números aleatorios utilizando la función default_rng() de NumPy.
rng = np.random.default_rng()

#Usamos para generar números enteros aleatorios entre 0 y 100, con un tamaño de 10.
a_values = rng.integers(low = 0, high = 100, size = 10) 

#Usamos para generar números enteros aleatorios entre 100 y 500, con un tamaño de 10.
b_values = rng.integers(low = 100, high = 500, size = 10)

#Creamos un DataFrame de pandas con dos columnas, 'A' y 'B', utilizando los valores generados anteriormente.
data = pd.DataFrame({'A': a_values, 'B': b_values})
data
```

```python
data.A.describe()
```

```python
#Definimos los intervalos para categorizar los valores de la columna 'A'.
bins = [0, 30, 40, 50, 60, 100]

#Usamos cut para categorizar los valores de la columna 'A' en función de los intervalos definidos en bins.
A_categories = pd.cut(data.A, bins, right= False)
A_categories.dtype
```

```python
#Creamos etiquetas para cada categoría definida por los intervalos en bins.
labels = ['Muy poco', 'Poco', 'Normal', 'Mucho', 'Demasiado']
A_categories_labels = pd.cut(data.A, bins, labels = labels)
A_categories_labels.value_counts()
```

```python
#Definimos una lista de categorías para asignar a la columna 'Category' del DataFrame.
categories = ['gato', 'perro', 'tigre', 'elefante', 'leon', 'raton']

#Asignamos aleatoriamente una categoría de la lista categories a cada fila del DataFrame, utilizando el generador de números aleatorios rng.
data['Category']= rng.choice(categories, size = 10)

#Asignamos aleatoriamente una categoría de tamaño ('pequeño', 'mediano', 'grande') 
# a cada fila del DataFrame, utilizando el generador de números aleatorios rng.
data['Size']= rng.choice(['pequeño', 'mediano', 'grande'], size = 10)

#Asignamos aleatoriamente un color ('Rojo', 'Azul', 'Verde') a cada fila del DataFrame, 
# utilizando el generador de números aleatorios rng.
data['Color'] = rng.choice(['Rojo', 'Azul', 'Verde'], size = 10)

print(data)
```

```python
#Creamos una instancia de LabelEncoder para codificar la columna 'Category' del DataFrame.
label_encoder = LabelEncoder()

#Utilizamos el método fit_transform del LabelEncoder para codificar la columna 'Category' 
# y asignamos el resultado a una nueva columna 'Category_encoded' en el DataFrame.  
data['Category_encoded'] = label_encoder.fit_transform(data['Category'])
print(data)
```

```python
#Creamos una instancia de OrdinalEncoder para codificar la columna 'Size' del DataFrame, 
# especificando el orden de las categorías en la lista categories.
ordinal_encoder = OrdinalEncoder(categories = [['pequeño', 'mediano', 'grande']])

#Utilizamos el método fit_transform del OrdinalEncoder para codificar la columna 'Size' 
# y asignamos el resultado a una nueva columna 'Size_encoded' en el DataFrame.
data['Size_encoded'] = ordinal_encoder.fit_transform(data[['Size']])
print(data)
```

```python
data = pd.DataFrame({'Category': ['gato', 'perro' , 'pato', 'gato', 'perro', 'raton']})
print(data)
```

```python
#Sparse es usado para indicar si la salida de la codificación debe ser una matriz dispersa (sparse) o una matriz densa (dense).
oneHencoder = OneHotEncoder(sparse_output = False)

oneHencoded = oneHencoder.fit_transform(data[['Category']])

oneHencoded_df = pd.DataFrame(oneHencoded, columns = oneHencoder.get_feature_names_out(['Category']))
print(oneHencoded_df)
```

```python
data_combined = pd.concat([data, oneHencoded_df], axis = 1)

print(data_combined)
```
